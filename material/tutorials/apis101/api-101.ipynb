## Before you start

This tutorial uses [music-to-scrape.org](https://music-to-scrape.org), a fictitious music streaming service used earlier in the course. Its [API documentation](https://api.music-to-scrape.org/docs) describes the available endpoints and parameters.

An API is a structured way for one program to request data from another program. Most APIs return JSON, which Python can read as dictionaries and lists.

| API part | Example | What it means |
|---|---|---|
| Endpoint | `/users` | The resource you want to request |
| Parameter | `limit=25` | An instruction that changes the response |
| Response | JSON | The structured data returned by the API |
| Record | `{"username": "StarCoder49"}` | One item in the response |

<div class="alert alert-block alert-info"><b>Beginner tip:</b><br>When an API response looks confusing, do not guess its structure. Print a small part of it, inspect its keys, and then extract one field at a time.</div>

<details>
<summary><b>For advanced users: why use an API instead of scraping?</b></summary>

If a suitable API exists, it is usually more stable and easier to parse than HTML. APIs also make parameters, pagination, and response structure explicit. Browser automation or web scraping is still useful when the data is not exposed through an API.

</details>

We will use the `/users` endpoint to retrieve user records. First, run the request and inspect the response before trying to extract anything from it.

**Checkpoint:** after running the cell, look for the `data`, `limit`, and `offset` keys in the response.


````python
# Import the 'requests' library, which is used to make HTTP requests.
import requests

# Define the URL of the API endpoint you want to request data from.
url = "https://api.music-to-scrape.org/users"

# Send an HTTP GET request to the specified URL and store the response in 'response'.
response = requests.get(url)

# Parse the JSON content of the response into a Python dictionary and store it in 'request'.
request = response.json()

# Inspect the response without flooding the notebook with all records.
print('Response keys:', request.keys())
print('Number of records:', len(request['data']))
print('First record:', request['data'][0])
````

    # Convert the modified 'username' dictionary to a JSON-formatted string and write it to the file.
    f.write(json.dumps(username))
    
    # Write a newline character ('\n') to separate each JSON object on a new line in the file.
